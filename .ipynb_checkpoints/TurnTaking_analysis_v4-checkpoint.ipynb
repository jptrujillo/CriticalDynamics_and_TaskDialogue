<h3>Introduction</h3>
Turn-taking is a fundamental aspect of human communication, and takes place quite rapidly, with turn transitions (i.e., the time between one speaker's utterance ending and the next speaker's utterance beginning) often on the scale of ~200ms (Stivers et al., doi:10.1073pnas.0903616106%20pna/). This timing is important not only for fast-pace information exchange, but is socially meaningful, as longer-than-expected inter-speaker gaps can be interpreted as signalling a dispreferred response (Kendrick & Torreira 2015  doi: 10.1080/0163853X.2014.955997). <br>
Beyond theoretical interest, quantifying turn-taking behavior can be interesting for understanding group dynamics. One study suggests that smoother turn transitions (i.e., shorter gaps, shorter overlaps) are more associated with collaborative behaviors compared to conflict behaviors (Kim et al., 2015, INTERSPEECH). Similarly, Pincus & Guastello (2005; DOI: 10.1177/1046496405280864) used non-linear dynamics approaches to show that information theoretic patterns in turn-taking behavior correlate with self-report measures of group dynamics (e.g., leadership, conflict). More recently, Hamish et al., (2023: https://escholarship.org/uc/item/2hb5k9fx ) have examined turn-taking entropy and critical instability (also described in this module) in relation to collaborative problem solving.

```python
 
def get_multispeaker_transitions(utterances_1, utterances_2, utterances_3):
    # set max distance between them to be 2s, then find the closest occurence in
    # second dataframe. Max distance is based on Heldner & Edlund 2010
    # output times are in ms
    
    speaker_count = 3
    if not utterances_1:
        speaker_count -=1
    if not utterances_2:
        speaker_count -=1
    if not utterances_3:
        speaker_count -=1
    
    if speaker_count < 2: 
        trans_dur = np.nan
        gap_dur = np.nan
        utterance_time_trans = np.nan
        utterance_time_gap = np.nan
    else:
        trans_dur1, gap_dur1, utterance_time_trans1, utterance_time_gap1, overlap_dur1, utterance_time_overlap1,timestamp1 = get_gaps_overlaps(
          utterances_1, utterances_2, utterances_3)
        trans_dur2, gap_dur2, utterance_time_trans2, utterance_time_gap2, overlap_dur2, utterance_time_overlap2,timestamp2 = get_gaps_overlaps(
          utterances_2, utterances_1, utterances_3) 
        trans_dur3, gap_dur3, utterance_time_trans3, utterance_time_gap3, overlap_dur3, utterance_time_overlap3,timestamp3 = get_gaps_overlaps(
          utterances_3, utterances_2, utterances_1) 
        
    
        #print("Turn Transition Statistics for Participant 1")
        #get_turn_trans_stats(trans_dur1,gap_dur1, overlap_dur1)
       # print("Turn Transition Statistics for Participant 2")
       # get_turn_trans_stats(trans_dur2,gap_dur2, overlap_dur2)
       # print("Turn Transition Statistics for Participant 3")
       # get_turn_trans_stats(trans_dur3,gap_dur3, overlap_dur3)
        
        # these will be used in the dataframe to indicate which speaker we're talking about
        p1 = ["1"]*len(trans_dur1)
        p2 = ["2"]*len(trans_dur2)
        p3 = ["3"]*len(trans_dur3)
        
        participant = p1 + p2 + p3
        timestamp = timestamp1 + timestamp2 + timestamp3

        # now we need to stack the utterance data
        trans_dur = trans_dur1 + trans_dur2 + trans_dur3
        utterance_time_trans = utterance_time_trans1 + utterance_time_trans2 + utterance_time_trans3
        
        df = pd.DataFrame(np.column_stack([participant,timestamp,trans_dur, utterance_time_trans]),
                               columns=['participant', 'timestamp','transition_duration','utterance_duration']
                               )
        
        return df
        
        
def get_gaps_overlaps(utterances_1, utterances_2, utterances_3):
    # this gets the utterance and turn-transition speaker from the perspective of one speaker
    # It doesn't matter what order the second two are in, as we just care about the averages in 
    # the end anyway
    trans_dur = []
    utterance_time_trans = []
    timestamp=[]

    for u_idx,u in enumerate(utterances_1):
        # here we want to get the difference between this utterance's offset (u[1]) and all onsets from the other two speakers
        diffs2 = [abs(float(u[1]) - float(u2[0])) for u2 in utterances_2]
        diffs3 = [abs(float(u[1]) - float(u3[0])) for u3 in utterances_3]

        min_dif2 = np.argmin(diffs2)
        min_dif3 = np.argmin(diffs3)        

        # also get within-speaker differences
        # we get this to make sure our inter-speaker differences are less than this
        # otherwise, we may not be getting the turn transition from this utterance
        if u_idx < len(utterances_1)-1:
            min_pause = utterances_1[u_idx+1][0] - u[1]
        else:
            # if this is not the last utterance, we ignore this
            min_pause = min([diffs2[min_dif2],diffs3[min_dif3]]) +1
        # is there an utterance from p2 or p3 starting within 2s of this utterance from p1?

        if diffs2[min_dif2] < 2000 and diffs2[min_dif2] < min_pause:
            trans_dur.append(int(utterances_2[min_dif2][0] - u[1]))
            utterance_time_trans.append(int(utterances_1[u_idx][1] - utterances_1[u_idx][0]))
            timestamp.append(int(utterances_2[min_dif2][0]))
        if diffs3[min_dif3] < 2000 and diffs3[min_dif3] < min_pause:
            trans_dur.append(int(utterances_3[min_dif3][0] - u[1]))
            utterance_time_trans.append(int(utterances_1[u_idx][1] - utterances_1[u_idx][0]))
            timestamp.append(int(utterances_3[min_dif3][0]))
            
    # finally, we want to separate these values into gaps and overlaps
    gap_dur = []
    overlap_dur = []
    utterance_time_gap = []
    utterance_time_overlap = []
    for idx, val in enumerate(trans_dur):
        if val > 0:
            gap_dur.append(val)
            utterance_time_gap.append(int(utterance_time_trans[idx]))
        elif val <= 0:
            overlap_dur.append(val)
            utterance_time_overlap.append(int(utterance_time_trans[idx]))

    return trans_dur, gap_dur, utterance_time_trans, utterance_time_gap, overlap_dur, utterance_time_overlap,timestamp
        
```

This function just gets us some summary stats for the turn taking behavior

```python
def get_turn_trans_stats(transitions, gaps, overlaps):
    if transitions and not np.all(np.isnan(transitions)):
        Gap_freq = len(gaps)/len(transitions)
        Overlap_freq = len(overlaps)/len(transitions)
    else:
        Gap_freq = 0
        Overlap_freq = 0
    turn_data = [['turn transition mean', np.mean(transitions)],
                 ['turn transition SD', np.std(transitions)],
                 ['turn transition median',np.median(transitions)],
                 ['Gap mean', np.mean(gaps)],
                 ['Gap SD',np.std(gaps)],
                 ['Gap median',  np.median(gaps)],
                 ['Gap freq. (%)',Gap_freq],
                 ['Overlap mean', np.mean(overlaps)],
                 ['Overlap SD',np.std(overlaps)],
                 ['Overlap median',  np.median(overlaps)],
                 ['Overlap freq. (%)',Overlap_freq],]
    
    print(tabulate(turn_data, headers=['Stat','Value']))
```

This part of the code is separated, because I'll be cleaning up the data in a way that is more consistent across the RQs. This way, the data going into each analyses is preprocessed in the same way. 

For the first RQ, we will get the speaker transcripts, remove intraspeaker transitions, then calculate turn-taking timing. 

```python
import antropy as ant
```

```python
%%capture
import scipy.stats

CPS = pd.read_csv("./annotations/All_Groups_CPS.csv")

# first get the unique group IDs
groups = list(set(CPS["Group"]))
groups.sort()

group_entropy = []

# for each of our groups..
for group in groups:
    transcript = pd.read_csv("./transcripts/" + group + "_Oracle.csv")
    utterances_1 = transcript[transcript["Participant"] == 1]
    utterances_1_SE = [(utterance[1]["Start"]*1000,utterance[1]["End"]*1000) for utterance in utterances_1.iterrows()]
    utterances_2 = transcript[transcript["Participant"] == 2]
    utterances_2_SE = [(utterance[1]["Start"]*1000,utterance[1]["End"]*1000) for utterance in utterances_2.iterrows()]
    utterances_3 = transcript[transcript["Participant"] == 3]
    utterances_3_SE = [(utterance[1]["Start"]*1000,utterance[1]["End"]*1000) for utterance in utterances_3.iterrows()]
    
    # ..get the turn-transition dataframes..
    df_cps = get_multispeaker_transitions(utterances_1_SE, utterances_2_SE, utterances_3_SE)
    
    # .. and then calculate entropy
    # first we need to sort this into a proper timeseries, disregarding individual IDs
    df_cps.timestamp = pd.to_numeric(df_cps.timestamp, errors='coerce')
    df_cps.transition_duration = pd.to_numeric(df_cps.transition_duration, errors='coerce')

    sort_df = df_cps.sort_values(by=["timestamp"], ascending=True)

    # Calculate Approximate entropy
    entropy = ant.sample_entropy(sort_df["transition_duration"])
    group_entropy.append(entropy)

df_cps_full = pd.DataFrame(list(zip(groups, group_entropy)),
                          columns = ["group","entropy"])
```

```python
#%%capture

group_instability_peaks = []
group_instability_peaks_idx = []
group_instability_peak_prom = []
group_instability_first10 = []

group_list_check = []
group_instability_mag = []
group_instability_prom = []

tt_entropy_full = []
tt_entropy_starts = []
# for each of our groups..
for group in groups:
    transcript = pd.read_csv("./transcripts/" +group + "_Oracle.csv")
    utterances_1 = transcript[transcript["Participant"] == 1]
    utterances_1_SE = [(utterance[1]["Start"]*1000,utterance[1]["End"]*1000) for utterance in utterances_1.iterrows()]
    utterances_2 = transcript[transcript["Participant"] == 2]
    utterances_2_SE = [(utterance[1]["Start"]*1000,utterance[1]["End"]*1000) for utterance in utterances_2.iterrows()]
    utterances_3 = transcript[transcript["Participant"] == 3]
    utterances_3_SE = [(utterance[1]["Start"]*1000,utterance[1]["End"]*1000) for utterance in utterances_3.iterrows()]
    
    # ..get the turn-transition dataframes..
    df_cps = get_multispeaker_transitions(utterances_1_SE, utterances_2_SE, utterances_3_SE)
    
    # .. and then calculate entropy
    # first we need to sort this into a proper timeseries, disregarding individual IDs
    df_cps.timestamp = pd.to_numeric(df_cps.timestamp, errors='coerce')
    df_cps.transition_duration = pd.to_numeric(df_cps.transition_duration, errors='coerce')

    sort_df = df_cps.sort_values(by=["timestamp"], ascending=True)
    
    tt_entropy = []
    tt_entropy_starts_local = []

    window_length = (1000*60) #written out just to make the window length easily understandable

    window_start =sort_df["timestamp"].loc[0]
    window_end =  window_start + window_length
    while window_end < sort_df["timestamp"][len(sort_df)-1]   :
        # get data for this time window
        window = sort_df[(sort_df["timestamp"] >window_start)&(sort_df["timestamp"] <window_end)]
        if len(window) > 3:
        # calculate permutation entropy
            tt_entropy.append(ant.perm_entropy(window["transition_duration"], normalize=True))
        else:
            tt_entropy.append(np.nan)
        # shift the window forward by 30s
        window_start = window_end - window_length/2 
        window_end = window_start + window_length

    tt_entropy_starts.append(tt_entropy_starts_local)
    tt_entropy_full.append(tt_entropy)
    instability_peaks, instability_prominence = signal.find_peaks(tt_entropy, height=.4, prominence=0.2, distance=2)

    group_list_check.append(group)
    group_instability_peaks.append(len(instability_peaks))
    group_instability_peaks_idx.append(instability_peaks)
    group_instability_peak_prom.append(instability_prominence["peak_heights"])
    group_instability_mag.append(np.mean(instability_prominence["peak_heights"]))
    group_instability_prom.append(np.mean(instability_prominence["prominences"]))
```

Because these values are collected in separate loops, we also collect the group ID for both lists so we can double check that the values are being correctly matched.

```python
time_on_task = []

for group in groups:
    transcript = pd.read_csv("./transcripts/" + group + "_Oracle.csv")
    ToT = transcript["End"].iloc[-1]
    time_on_task.append(ToT)
    
df_cps_full["ToT"] = time_on_task  
```

In team-based problem-solving tasks, moments of critical instability may be necessary changes in the dynamic of a conversation/discussion, where a team is (for example) changing strategies or trying to find a stable dynamic. Accordingly, many such changes in dynamics are likely to be associated with a longer time working on a problem. Teams that find a stable, effective dynamic with fewer such changes, will then be faster. Even without looking into what exactly is occurring during a conversation, this type of analysis can provide a first look at whether moments of critical instability are associated with how long it takes to complete a task.<br>


```python
fit = np.polyfit(df_cps_full["ToT"], group_instability_peaks,1, full=True)
slope=fit[0][0]
intercept=fit[0][1]
xl = [min(df_cps_full["ToT"]), max(df_cps_full["ToT"])]
yl = [slope*xx + intercept  for xx in xl]


plt.scatter(df_cps_full["ToT"],group_instability_peaks)
plt.plot(xl, yl, '-r')
plt.title("Association Between Time-on-Task and Number of Instability Peaks")
plt.show()
```

This is a nice check, but instability peaks are likely to occur as time goes on, so there's likely a positive association between number of peaks and length of time-series, regardless of anything function. So we need to take the <i>rate</i> of peaks/phase transitions to factor this out. 


```python
peaks_per_min = group_instability_peaks/df_cps_full["ToT"]*60
fit = np.polyfit(df_cps_full["ToT"], peaks_per_min,1, full=True)
slope=fit[0][0]
intercept=fit[0][1]
xl = [min(df_cps_full["ToT"]), max(df_cps_full["ToT"])]
yl = [slope*xx + intercept  for xx in xl]


plt.scatter(df_cps_full["ToT"],peaks_per_min)
plt.plot(xl, yl, '-r')
plt.title("Association Between Time-on-Task and Number of Rate of Instbality Peaks")
plt.show()
```

```python
import scipy.stats
```

```python
res = scipy.stats.pearsonr(df_cps_full["ToT"], peaks_per_min)
rng = np.random.default_rng()

method = scipy.stats.BootstrapMethod(method='BCa')
print(res)
print(res.confidence_interval(confidence_level=0.9, method=method))
```

<b>Result:</b> Higher rate of phase transitions/instability peaks is indeed associated with longer time on task, as predicted

<b>SubQuestion: Does the magnitude of critical instability points correlate with group performance?</b><br>

Following Wiltshire et al., 2021, let's also check if the average magnitude of critical instability points is associated with time-on-task.

```python
nan_list = np.where(np.isnan(group_instability_mag))
print(nan_list)
```

```python

ToT_clean = df_cps_full["ToT"]
group_instability_mag_clean = group_instability_mag.copy()

for nan_idx in nan_list[0]:
    nan_idx = int(nan_idx)
    ToT_clean.pop(nan_idx)
    group_instability_mag_clean.pop(nan_idx)

res = scipy.stats.pearsonr(ToT_clean, group_instability_mag_clean)

rng = np.random.default_rng()

method = scipy.stats.BootstrapMethod(method='BCa')
print(res)
print(res.confidence_interval(confidence_level=0.9, method=method))
```

<b>Result:</b> We find no evidence for the <i>magnitude</i> of instability peaks correlating with time on task.

### Question 2: Modeling Information Flow Around Transition Points

Here, we'll use the Flow Score (basically, semantic influence) to quantify the flow of information, or semantic coupling, between speaking turns. The main question here is: are moments of critical instability preceded by, or followed by, changes in semantic flow? <br>
This is based on Burdisso, 2024's DSE (as implemented in Dialog2Flow) model (https://huggingface.co/sergioburdisso/dialog2flow-joint-dse-base). <br>
What this is actually doing is calculating the semantic similarity between each pair of consecutive turns. The model is trained on dialogue data, so it should have a better representation of turn-to-turn similarity than, for example, a more basic word embedding model that is just treating each utterance as an independent bag of words. 

First, we need to load the model. Then, preprocess the transcripts a bit. Then, we calculate Flow.

<b>NOTE: the semantic flow values have been saved to a pickle container, so the next few code blocks can be skipped in order to not re-run the model

```python
from sentence_transformers import SentenceTransformer
from transformers import AutoTokenizer, AutoModel
import torch
from numpy import dot
from numpy.linalg import norm

#Mean Pooling - Take attention mask into account for correct averaging
def mean_pooling(model_output, attention_mask):
    token_embeddings = model_output[0] #First element of model_output contains all token embeddings
    input_mask_expanded = attention_mask.unsqueeze(-1).expand(token_embeddings.size()).float()
    return torch.sum(token_embeddings * input_mask_expanded, 1) / torch.clamp(input_mask_expanded.sum(1), min=1e-9)


def cosine_similarity(a,b):
    cos_sim = dot(a, b)/(norm(a)*norm(b))

    return cos_sim

# Sentences we want sentence embeddings for
#sentences = ['your phone please', 'okay may i have your telephone number please']
#sentences = ['What do we do next', 'okay may i have your telephone number please']
#sentences = ['What do we do next', 'We can try this other thing']

# Load model from HuggingFace Hub
tokenizer = AutoTokenizer.from_pretrained('sergioburdisso/dialog2flow-joint-dse-base')
model = AutoModel.from_pretrained('sergioburdisso/dialog2flow-joint-dse-base')
```

First, we need to preprocess the utterances, similarly to above, but now maintaining the transcript utterances, while removing intraspeaker gaps. 

```python
CPS = pd.read_csv("./annotations/All_Groups_CPS.csv")

# first get the unique group IDs
groups = list(set(CPS["Group"]))
groups.sort()

df_full = []

for group in groups:
    utterance =  []
    start_time = []
    end_time = []
    speaker = []
    
    transcript = pd.read_csv("./transcripts/" + group + "_Oracle.csv")

    current_speaker = transcript["Participant"][0]
    current_utterance = transcript["Transcript"][0]
    current_start = transcript["Start"][0]
    current_end = transcript["End"][0]
    
    for index, row in transcript.iterrows():
        if index > 0:
            if row["Participant"] == current_speaker or index == len(transcript)<1:
                extension = " " + row["Transcript"]
                current_utterance += extension
                current_end = row["End"]
            else:
                utterance.append(current_utterance)
                start_time.append(current_start)
                speaker.append(current_speaker)
                end_time.append(current_end)

                current_speaker = row["Participant"]
                current_utterance = row["Transcript"]
                current_start = row["Start"]
                current_end = row["End"]

    
    df_group = pd.DataFrame(np.column_stack([start_time, end_time, speaker, utterance]),
                               columns=["start","end", "speaker", "utterance"],
                               )
    df_full.append(df_group)


    
    
```

```python
CPS = pd.read_csv("./annotations/All_Groups_CPS.csv")

# first get the unique group IDs
groups = set(CPS["Group"])

df_full = []

for group in groups:
    utterance =  []
    start_time = []
    end_time = []
    speaker = []
    
    transcript = pd.read_csv("./transcripts/" + group + "_Oracle.csv")

    current_speaker = transcript["Participant"][0]
    current_utterance = transcript["Transcript"][0]
    current_start = transcript["Start"][0]
    current_end = transcript["End"][0]
    
    for index, row in transcript.iterrows():
        if index > 0:
        
            utterance.append(current_utterance)
            start_time.append(current_start)
            speaker.append(current_speaker)
            end_time.append(current_end)

            current_speaker = row["Participant"]
            current_utterance = row["Transcript"]
            current_start = row["Start"]
            current_end = row["End"]

    
    df_group = pd.DataFrame(np.column_stack([start_time, end_time, speaker, utterance]),
                               columns=["start","end", "speaker", "utterance"],
                               )
    df_full.append(df_group)
```

Now we can encode, calculate similarity, calculate entropy

```python

sim_full = []

for df_group in df_full:          
    sim_list = []
    start_list = []

    sentences = list(df_group["utterance"])
    #Tokenize sentences
    encoded_input = tokenizer(sentences, padding=True, truncation=True, return_tensors='pt')

    # Compute token embeddings
    with torch.no_grad():
        model_output = model(**encoded_input)

    # Perform pooling. In this case, mean pooling.
    sentence_embeddings = mean_pooling(model_output, encoded_input['attention_mask'])

    for index in range(1,len(sentence_embeddings)):
        sim = cosine_similarity(sentence_embeddings[index-1],sentence_embeddings[index])
        sim_list.append(sim)
        start_list.append(df_group["start"][index-1])

    sim_full.append((start_list, sim_list))
    
```

After all this, we can calculate entropy using the same sliding time-window approach as we did for turn-taking, and bring together with turn-taking data to visualize how the patterns line up.

```python
import antropy as ant
import scipy.stats
```

```python
sim_entropy_full = []
sim_entropy_starts_full = []
# for each of our groups..
for group in sim_full:

    timestamps = [float(val) for val in group[0]]
    sim_scores = [float(val) for val in group[1]]
    
    # .. and then calculate entropy
    sim_entropy = []
    sim_entropy_starts_local = []
    
    window_length = 30 #written out just to make the window length easily understandable

    window_start = 0
    #window_start = float(timestamps[0])
    window_end =  window_start + window_length
    
    while window_end < timestamps[len(timestamps)-1]:
        sim_entropy_starts_local.append(window_start)
        # get data for this time window
        
        window_indices = [idx for idx,x in enumerate(timestamps) if (timestamps[idx] >= window_start) & (timestamps[idx] < window_end)]
        window = [sim_scores[idx] for idx in window_indices]
        
        if len(window) > 3:
        # calculate permutation entropy
            sim_entropy.append(ant.perm_entropy(window, normalize=True))
        else:
            sim_entropy.append(np.nan)
        # shift the window forward by 30s
        window_start = window_end - window_length/2 
        window_end = window_start + window_length

    sim_entropy_full.append(sim_entropy)
    sim_entropy_starts_full.append(sim_entropy_starts_local)
```

```python
# this is to trim off the last bit of the semantic flow entropy, when there's a mismatch. 
# This is just because the windows don't quite align

for group in range(len(sim_entropy_full)):
    #print(len(tt_entropy_full[group]), (len(sim_entropy_full[group])))
    
    if len(sim_entropy_full[group]) > len(tt_entropy_full[group]):
        sim_entropy_full[group] = sim_entropy_full[group][0:-1]
    #print(len(tt_entropy_full[group]), (len(sim_entropy_full[group])))

```

Next, we formally test whether semantic entropy close to the TT entropy peaks is different from baseline.

<b>SubQ 1: is semantic flow immediately before phase transition different than average? </b><br>
The hypothesis here is that semantic flow will be lower, representing divergence immediately before a transition. The method: 1) calculate the global mean of semantic flow for each group, 2) take a 2min window before each instability peak, 3) calculate mean semantic flow within this window, 4) use mixed regression to test difference from group mean. <br>
Basically, we're testing if the semantic entropy immediately before or after a turn-taking instability peak are higher than the average semantic entropy for that particular group.

```python
def fwhm2std(fwhm):
    """from fwhm to standard deviation of a gaussian"""
    return fwhm/2.355

def smoothing(data, window_size=5):
    data = pd.Series(data).interpolate(method="spline",order=2)
    
    #window_size = 5
    
    std = fwhm2std(window_size)
    smoothed_time_series = data.rolling(window_size, win_type="gaussian", center=True).mean(std=std)

    return smoothed_time_series
```

```python
lead_values = []
follow_values = []
group_list = []
peak_list = []
lead_diff_values = []
follow_diff_values = []

tt_entropy_smooth = []

group_means = []
group_sd = []
group_ci = []

def calc_ci(data):

    data_int = pd.Series(data).interpolate(method="spline",order=2)
    
    ci = scipy.stats.t.interval(confidence=0.99, df=len(data_int)-1, 
                                loc=np.mean(data_int), 
                                scale=scipy.stats.sem(data_int, nan_policy="omit")) 
    
    return ci


for group in range(len(tt_entropy_full)):
    current_peaks = group_instability_peaks_idx[group]
    current_sim_entropy = sim_entropy_full[group]

    tt_entropy_smooth.append(smoothing(tt_entropy_full[group],window_size=3))

    group_means.append(np.nanmean(tt_entropy_full[group]))
    group_sd.append(np.nanstd(tt_entropy_full[group]))
    group_ci.append(calc_ci(tt_entropy_smooth[group]))

        
    peak_idx = 0
    for peak in current_peaks:
        if peak > 6:
            # take the preceding 4 data points
            lead_sim = current_sim_entropy[peak-6:peak]
            follow_sim = current_sim_entropy[peak+1:peak+7]
            
            follow_values.extend(follow_sim)
            lead_values.extend(lead_sim)
            
            follow_diff_values.extend([val-group_means[group] for val in follow_sim])
            lead_diff_values.extend([val-group_means[group] for val in lead_sim])
            
            group_list.extend([group]*len(lead_sim))
            peak_list.extend([peak_idx]*len(lead_sim))
    
            peak_idx+=1
            #if len(follow_values) != len(group_list):
            #    sys.exit()

df_entropy = pd.DataFrame(np.column_stack([group_list, peak_list, lead_values, follow_values, lead_diff_values, follow_diff_values]),
                               columns=["group","peak", "lead", "follow", "lead_diff", "follow_diff"],
                               )
```

```python
import random

def generate_pseudo_df():
    lead_values = []
    follow_values = []
    group_list = []
    peak_list = []
    lead_diff_values = []
    follow_diff_values = []
    
    group_means = []
    
    for group in range(len(tt_entropy_full)):
        current_peaks = group_instability_peaks_idx[group]
        current_sim_entropy = sim_entropy_full[group]
    
        group_means.append(np.nanmean(tt_entropy_full[group]))
    
        peak_idx = 0
        for peak in current_peaks:
            pseudo_peak = random.randint(5,len(current_sim_entropy)-5)
            # take the preceding 4 data points
            lead_sim = current_sim_entropy[pseudo_peak-4:pseudo_peak]
            follow_sim = current_sim_entropy[pseudo_peak+1:pseudo_peak+5]
            lead_values.extend(lead_sim)
            follow_values.extend(follow_sim)

            follow_diff_values.extend([val-group_means[group] for val in follow_sim])
            lead_diff_values.extend([val-group_means[group] for val in lead_sim])
            
            group_list.extend([group]*len(lead_sim))
            peak_list.extend([peak_idx]*len(lead_sim))
    
            peak_idx+=1
    
    df_entropy_pseudo = pd.DataFrame(np.column_stack([group_list, peak_list, lead_values, follow_values]),
                                   columns=["group","peak", "lead", "follow"],
                                   )
    return df_entropy_pseudo
```

### Question 3: Is gesture use associated with faster reorganization?
This is based on the idea that gestures regulate turn-taking in some ways, and also help to communicate more generally. <br>
In terms of reorganization, we can use Grimm's (2023) operationalization as a time to return to baseline: https://journals.sagepub.com/doi/full/10.1177/15553434231199729 <br>
Specifically, we'll use the upper bound of the 99% Confidence Interval of the data. For this to make sense, we'll first interpolate the missing data, and then smooth it. <br>
Gesture rate during return-to-baseline window can be correlated with reorganization time.
<br>



First, we need to calculate the reorganization windows, then extract all gestures from that window.<br>
Data: <br>
- the peaks are defined in group_instability_peaks_idx
- the time-series data are in tt_entropy_full

```python
gesture_annot_files = [filename for filename in os.listdir("./annotations/") if "GAMR" in filename and filename.endswith(".eaf")]
print(gesture_annot_files)
```

```python
def get_window_of_annotations(eafob, group,start_time, end_time):
    # requires a bit of exception handling due to somevariability in naming conventions
    if group ==5:
        gesture_annots = [annotation for annotation in eafob.get_annotation_data_between_times("Tier-1", start_time, end_time)]
    else:
        try:
            g_1 = [annotation for annotation in eafob.get_annotation_data_between_times("Participant 1", start_time, end_time)]
            g_2 = [annotation for annotation in eafob.get_annotation_data_between_times("Participant 2", start_time, end_time)]
            g_3 = [annotation for annotation in eafob.get_annotation_data_between_times("Participant 3", start_time, end_time)]
        except:
            g_1 = [annotation for annotation in eafob.get_annotation_data_between_times("Participant_1", start_time, end_time)]
            g_2 = [annotation for annotation in eafob.get_annotation_data_between_times("Participant_2", start_time, end_time)]
            g_3 = [annotation for annotation in eafob.get_annotation_data_between_times("Participant_3", start_time, end_time)]
        gesture_annots = g_1 + g_2 + g_3

    return gesture_annots



group_list = []
peak_idx_list =[]
relax_dur_list = []
relax_gesture_list = []
relax_gest_rate_list = [] # rate is avg number of gestures per second
deixis_count_list = []
emblem_count_list = []
iconic_count_list = []
sem_entropy_list = []
sem_entropy_slope_list = []

for group in range(len(group_instability_peaks_idx)-1):
    for peak_idx in group_instability_peaks_idx[group]:
        # get the CI threshold
        threshold = group_ci[group][1]

        # get the gesture coding data
        eafob = pympi.Elan.Eaf("./annotations/" + gesture_annot_files[group])
        
        if peak_idx > threshold:
            
            time_series = tt_entropy_smooth[group][peak_idx:len(tt_entropy_smooth[group])]

            stop_idx = next(x for x,y in enumerate(time_series) if y < threshold and x >0)

            # duration, in milliseconds, of the relaxation period (return to baseline entropy)
            relax_window_dur_ms = window_length + (window_length/2*stop_idx)
            # the actual timestamps associated with the relaxation period
            start_time = peak_idx*window_length
            end_time = start_time + (stop_idx*window_length)
            gesture_annots = get_window_of_annotations(eafob, group,start_time, end_time) 

            # semantic entropy
            sem_entropy_window = sim_entropy_full[group][peak_idx:peak_idx+stop_idx+1]
            time_ts = range(len(sem_entropy_window))
            slope, _, _, _, _ = scipy.stats.linregress(time_ts,sem_entropy_window)
            mean_sem_entropy = np.nanmean(sem_entropy_window)

            # store the data
            group_list.append(group)
            peak_idx_list.append(peak_idx)
            relax_dur_list.append(relax_window_dur_ms/1000)
            relax_gesture_list.append(len(gesture_annots))
            relax_gest_rate_list.append((len(gesture_annots)/relax_window_dur_ms/1000)*60)
            deixis_count_list.append(len([gest for gest in gesture_annots if "deixis" in gest[2]]))
            emblem_count_list.append(len([gest for gest in gesture_annots if "emblem" in gest[2]]))
            iconic_count_list.append(len([gest for gest in gesture_annots if "iconic" in gest[2]]))
            sem_entropy_list.append(mean_sem_entropy)
            sem_entropy_slope_list.append(slope)
            
            
g_df = pd.DataFrame(list(zip(group_list, peak_idx_list, relax_dur_list, 
                             relax_gesture_list, relax_gest_rate_list, deixis_count_list, emblem_count_list, iconic_count_list,
                            sem_entropy_list, sem_entropy_slope_list)),
                    columns=["group", "peak_id", "relax_duration", 
                             "gesture_count", "gesture_rate", "deixis_count", "emblem_count", "iconic_count",
                            "mean_semantic_entropy", "slope_of_semantic_entropy"])
                             
```
